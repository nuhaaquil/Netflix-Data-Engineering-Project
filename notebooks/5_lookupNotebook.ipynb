# Run notebook on specific day- using workflow

### 1. Create parameter for day number

```python
dbutils.widgets.text("weekday","7")
```

### 2. Store the day number

```python
var = int(dbutils.widgets.get("weekday"))
```

### 3. Set the value using "key" and "value"

```python
dbutils.jobs.taskValues.set(key="weekoutput",value=var)
```
